### 1. MNIST数据集简介  
[MNIST 数据集](http://yann.lecun.com/exdb/mnist/)来自美国国家标准与技术研究所（National Institute of Standards and Technology，简称 NIST ），总共有7万张图，其中训练集6万张，由 250 个不同人的手写数字构成, 50% 是高中学生, 另外 50% 是人口普查局的工作人员，测试集1万张图，也是由同样比例的人手写的数字，每张图片都是28\*28大小的黑白图像。

该数据集在深度学习领域，是一个很经典的入门学习数据集，部分手写数字的图片样例如下：  

![mnist](https://modelarts-labs-bj4-v2.obs.cn-north-4.myhuaweicloud.com:443/course/hwc_edu/deep_learning/datasets/imgs/mnist.jpg)

### 2. 下载MNIST数据集  
运行如下代码即可将数据集下载到当前datasets目录下的MNIST_data子目录中

```python
# 创建datasets目录
import os
datasets_dir = '../datasets'
if not os.path.exists(datasets_dir):
    os.makedirs(datasets_dir)
```

下载数据集，由于数据源在华为云OBS中，所以本代码只能在华为云 ModelArts 中运行

```python
if not os.path.exists(os.path.join(datasets_dir, 'MNIST_data.zip')):
    import moxing as mox
    mox.file.copy('obs://modelarts-labs-bj4-v2/course/hwc_edu/deep_learning/datasets/MNIST_data.zip', 
                  os.path.join(datasets_dir, 'MNIST_data.zip'))
    os.system('cd %s; unzip MNIST_data.zip' % (datasets_dir))
```

```python
import zipfile
f = zipfile.ZipFile(os.path.join(datasets_dir, 'MNIST_data.zip'))
for file in f.namelist():
    f.extract(file, datasets_dir)
f.close()
```

### 3. 读取MNIST数据集图片

```python
import os
import torchvision.datasets.mnist as mnist
```

```python
datasets_dir = '../datasets'
# 读取训练样本
train_data = mnist.read_image_file(os.path.join(datasets_dir, 'MNIST_data/raw/train-images-idx3-ubyte'))
train_label = mnist.read_label_file(os.path.join(datasets_dir, 'MNIST_data/raw/train-labels-idx1-ubyte'))
# 读取测试样本
test_data = mnist.read_image_file(os.path.join(datasets_dir, 'MNIST_data/raw/t10k-images-idx3-ubyte'))
test_label = mnist.read_label_file(os.path.join(datasets_dir, 'MNIST_data/raw/t10k-labels-idx1-ubyte'))
```

```python
print('训练集规模：', train_data.shape, train_label.shape)  # 60000个训练样本
print('训练集规模：', test_data.shape, test_label.shape)  # 10000个训练样本
```

### 4. 查看图片及其标签

```python
from PIL import Image
img = train_data[1].numpy()  # train_data[1]是tensor格式，转成numpy格式
print('图像的标签:', train_label[1].item())
print('图像的大小：', img.shape)
Image.fromarray(img)  # 转成PIL格式进行图片显示

for i in range(10):
    img = train_data[i].numpy()  # train_data[1]是tensor格式，转成numpy格式
    print('图像的标签:', train_label[i].item())
    print('图像的大小：', img.shape)
    Image.fromarray(img)  # 转成PIL格式进行图片显示
```

```python
img = train_data[30].numpy()  # train_data[0]是tensor格式，转成numpy格式
print('图像的标签:', train_label[30].item())
print('图像的大小：', img.shape)
Image.fromarray(img)  # 转成PIL格式进行图片显示
```
